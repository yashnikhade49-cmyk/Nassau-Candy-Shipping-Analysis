```python
# Route Efficiency Score


import pandas as pd

# Load the final route analysis from Step 6
final_route_analysis = pd.read_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\route_efficiency_analysis.csv",
    index_col=0
)

# Calculate minimum and maximum average lead time
min_lead_time = final_route_analysis["Average_Lead_Time"].min()
max_lead_time = final_route_analysis["Average_Lead_Time"].max()

# Calculate Route Efficiency Score
final_route_analysis["Efficiency_Score"] = (
    100 * (
        1 -
        (
            final_route_analysis["Average_Lead_Time"] - min_lead_time
        ) /
        (
            max_lead_time - min_lead_time
        )
    )
)

# Round the score for easier reading
final_route_analysis["Efficiency_Score"] = (
    final_route_analysis["Efficiency_Score"].round(2)
)

# Sort from most efficient to least efficient
final_route_analysis = final_route_analysis.sort_values(
    "Efficiency_Score",
    ascending=False
)

print("Route Efficiency Score:")
display(final_route_analysis.head(20))
```

```python
# Top 10 and Bottom 10 Routes

# Top 10 most efficient routes
top_10_routes = (
    final_route_analysis
    .sort_values("Efficiency_Score", ascending=False)
    .head(10)
)

print("Top 10 Most Efficient Routes: ")
display(top_10_routes[
    [
        "Shipment_Volume",
        "Average_Lead_Time",
        "Delayed_Shipments",
        "Delay_Frequency",
        "Efficiency_Score"
    ]
])

# Bottom 10 least efficient routes
bottom_10_routes = (
    final_route_analysis
    .sort_values("Efficiency_Score", ascending=True)
    .head(10)
)

print("Bottom 10 Least Efficient Routes:")
display(bottom_10_routes[
    [
        "Shipment_Volume",
        "Average_Lead_Time",
        "Delayed_Shipments",
        "Delay_Frequency",
        "Efficiency_Score"
    ]
])
```

```python
# Efficiency Score vs Delay Frequency

# Select important columns for comparison
efficiency_delay_comparison = final_route_analysis[
    [
        "Shipment_Volume",
        "Average_Lead_Time",
        "Delay_Frequency",
        "Efficiency_Score"
    ]
].copy()

# Sort by Efficiency Score
efficiency_delay_comparison = (
    efficiency_delay_comparison
    .sort_values("Efficiency_Score", ascending=False)
)

print("Efficiency Score vs Delay Frequency:")
display(efficiency_delay_comparison.head(20))
```

```python
# Calculate median benchmarks
efficiency_benchmark = final_route_analysis["Efficiency_Score"].median()
delay_benchmark = final_route_analysis["Delay_Frequency"].median()

print("Efficiency Score Benchmark:", efficiency_benchmark)
print("Delay Frequency Benchmark:", delay_benchmark)


# Create route performance classification
def classify_route(row):

    if (
        row["Efficiency_Score"] >= efficiency_benchmark
        and row["Delay_Frequency"] <= delay_benchmark
    ):
        return "Efficient & Low Delay"

    elif (
        row["Efficiency_Score"] >= efficiency_benchmark
        and row["Delay_Frequency"] > delay_benchmark
    ):
        return "Efficient but Delay-Prone"

    elif (
        row["Efficiency_Score"] < efficiency_benchmark
        and row["Delay_Frequency"] <= delay_benchmark
    ):
        return "Inefficient but Low Delay"

    else:
        return "Inefficient & Delay-Prone"


final_route_analysis["Route_Classification"] = (
    final_route_analysis.apply(
        classify_route,
        axis=1
    )
)


# Display classification results
print("Route Performance Classification:")
display(
    final_route_analysis[
        [
            "Shipment_Volume",
            "Average_Lead_Time",
            "Delay_Frequency",
            "Efficiency_Score",
            "Route_Classification"
        ]
    ].head(20)
)
```

```python
# Save the final Step 7 analysis
final_route_analysis.to_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\final_route_efficiency_analysis.csv",
    index=True
)

print("Final Step 7 route efficiency analysis saved successfully.")
```
