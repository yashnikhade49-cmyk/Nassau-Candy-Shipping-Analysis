```python
import pandas as pd

df_features = pd.read_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\factory_mapped_data.csv"
)
```

```python
df_features.head()
```

```python
# oandas read the dates as a string so when we have imported the csv file date is treated as a stirng. so, we are 
# using following program to convert it into actual date and time.

df_features["Order Date"] = pd.to_datetime(
    df_features["Order Date"], format="%d-%m-%Y"
)

df_features["Ship Date"] = pd.to_datetime(
    df_features["Ship Date"], format="%d-%m-%Y"
)
```

```python
# Calculating shipping lead time and storing it into the column
df_features["Shipping Lead Time"] = (df_features["Ship Date"] - df_features["Order Date"]).dt.days
```

```python
# Creating a factory customer state route
df_features["Route"] = (
    df_features["Factory"] + " → " + df_features["State/Province"]
)
```

```python
df_features["Region Route"] = (
    df_features["Factory"] + " → " + df_features["Region"]
)
```
