```python
import pandas as pd

# Load route efficiency analysis
route_analysis = pd.read_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\final_route_efficiency_analysis.csv",
    index_col=0
)

# Load geographic analysis
geo_analysis = pd.read_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\geographic_bottleneck_analysis.csv",
    index_col=0
)

# Load ship mode analysis
ship_mode_analysis = pd.read_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\ship_mode_performance_analysis.csv",
    index_col=0
)

print("All final analysis files loaded successfully.")
```

```python
# Routes with highest efficiency scores
top_routes = (
    route_analysis
    .sort_values("Efficiency_Score", ascending=False)
    .head(10)
)

# Routes with lowest efficiency scores
bottom_routes = (
    route_analysis
    .sort_values("Efficiency_Score", ascending=True)
    .head(10)
)

# Routes with highest delay frequency
delay_prone_routes = (
    route_analysis
    .sort_values("Delay_Frequency", ascending=False)
    .head(10)
)

print("Top 10 Routes by Efficiency Score:")
display(top_routes)

print("\nBottom 10 Routes by Efficiency Score:")
display(bottom_routes)

print("\nTop 10 Routes by Delay Frequency:")
display(delay_prone_routes)
```

```python
print("========== GEOGRAPHIC INSIGHTS ==========")

print("\nPotential Geographic Bottlenecks:")

display(
    geo_analysis[
        geo_analysis["Geographic_Classification"]
        == "Potential Geographic Bottleneck"
    ]
    .sort_values(
        ["Shipment_Volume", "Average_Lead_Time"],
        ascending=[False, False]
    )
)
```

```python
print("========== SHIP MODE INSIGHTS ==========")

ship_mode_insights = (
    ship_mode_analysis[
        [
            "Shipment_Volume",
            "Average_Lead_Time",
            "Average_Cost",
            "Average_Sales"
        ]
    ]
    .sort_values("Average_Lead_Time")
)

display(ship_mode_insights)
```

```python
# Create Final Insights Table

business_summary = pd.DataFrame({
    "Analysis Area": [
        "Route Efficiency",
        "Route Delays",
        "Geographic Performance",
        "Ship Mode Performance"
    ],
    "Key Metric": [
        route_analysis["Efficiency_Score"].max(),
        route_analysis["Delay_Frequency"].max(),
        geo_analysis["Average_Lead_Time"].max(),
        ship_mode_analysis["Average_Lead_Time"].min()
    ],
    "Interpretation": [
        "Highest observed route efficiency score",
        "Highest observed route delay frequency",
        "Highest observed state average lead time",
        "Lowest observed ship mode average lead time"
    ]
})

display(business_summary)
```

```python
business_summary.to_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\business_insights_summary.csv",
    index=False
)

print("Business insights summary saved successfully.")
```
