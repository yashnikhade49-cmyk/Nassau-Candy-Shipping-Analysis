```python
import pandas as pd

df_route = pd.read_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\feature_engineered_data.csv"
)

df_route.head()
```

```python
# Basic statistics of Shipping Lead Time
print("Shipping Lead Time Summary:")
display(df_route["Shipping Lead Time"].describe())

# Check minimum and maximum values
print("\nMinimum Lead Time:")
print(df_route["Shipping Lead Time"].min())

print("\nMaximum Lead Time:")
print(df_route["Shipping Lead Time"].max())

# Show the 10 largest lead times
print("\nTop 10 Largest Lead Times:")
display(
    df_route.nlargest(
        10,
        "Shipping Lead Time"
    )[["Order ID", "Order Date", "Ship Date", "Shipping Lead Time"]]
)
```

```python
# Route-wise shipment count
route_volume = (
    df_route.groupby("Route")
    .size()
    .sort_values(ascending=False)
)

print("Top 10 Routes by Shipment Volume:")
display(route_volume.head(10))


# Route-wise average shipping lead time
route_avg_lead = (
    df_route.groupby("Route")["Shipping Lead Time"]
    .mean()
    .sort_values()
)

print("Top 10 Fastest Routes:")
display(route_avg_lead.head(10))


print("Top 10 Slowest Routes:")
display(route_avg_lead.tail(10).sort_values(ascending=False))
```

```python
# Route Performance Summary
route_performance = (
    df_route.groupby("Route")
    .agg(
        Shipment_Volume = ("Route", "size"),
        Average_Lead_Time = ("Shipping Lead Time", "mean"),
        Minimum_Lead_Time = ("Shipping Lead Time", "min"),
        Maximum_Lead_Time = ("Shipping Lead Time", "max")
    )
    .sort_values("Shipment_Volume", ascending = False)
)

print("Route performance Summary")
display(route_performance.head(20))
```

```python
# Keep only routes with at least 10 shipments
reliable_routes = route_performance[
    route_performance["Shipment_Volume"] >=10
].copy()

print("Route with at least 10 shipments: ")
display(reliable_routes.head(20))
```

```python
# 10 fastest reliable routes
fastest_reliable = (
    reliable_routes
    .sort_values("Average_Lead_Time")
    .head(10)
)

print("10 Fastest Reliable Routes:")
display(fastest_reliable)


# 10 slowest reliable routes
slowest_reliable = (
    reliable_routes
    .sort_values("Average_Lead_Time", ascending=False)
    .head(10)
)

print("10 Slowest Reliable Routes:")
display(slowest_reliable)
```

```python
# Route dely Threshold
delay_threshold = 5

# Note:
# A 5-day threshold was initially tested.
# Since all shipments have Shipping Lead Time > 5 days,
# this threshold results in 100% delay frequency.
# Therefore, a data-driven threshold will be selected
# before final delay analysis.

# Create a delayed shipment indicator
df_route["Delayed"] = (
    df_route["Shipping Lead Time"] > delay_threshold
)
# Calculate route-wise delay statistics
route_delay = (
    df_route.groupby("Route")
    .agg(
        Shipment_Volume=("Route", "size"),
        Delayed_Shipments=("Delayed", "sum"),
        Delay_Frequency=("Delayed", "mean")
    )
)

# Convert delay frequency into percentage
route_delay["Delay_Frequency"] = (
    route_delay["Delay_Frequency"] * 100
)

# Sort routes by highest delay frequency
route_delay = route_delay.sort_values(
    "Delay_Frequency",
    ascending=False
)

print("Route Delay Frequency:")
display(route_delay.head(20))
```

```python
delay_percentiles = df_route["Shipping Lead Time"].quantile(
    [0.50, 0.75, 0.80, 0.90, 0.95]
)

print("Shipping Lead Time Percentiles:")
display(delay_percentiles)
```

```python
# Data-driven delay benchmark
delay_threshold = df_route["Shipping Lead Time"].quantile(0.75)

print("Data-driven delay threshold:", delay_threshold, "days")
```

```python
# Set the data-driven delay benchmark
delay_threshold = df_route["Shipping Lead Time"].quantile(0.75)

print("Delay Benchmark:", delay_threshold, "days")


# Create a delayed shipment indicator
df_route["Delayed"] = (
    df_route["Shipping Lead Time"] > delay_threshold
)


# Calculate route-wise delay statistics
route_delay = (
    df_route.groupby("Route")
    .agg(
        Shipment_Volume=("Route", "size"),
        Delayed_Shipments=("Delayed", "sum"),
        Delay_Frequency=("Delayed", "mean")
    )
)


# Convert delay frequency into percentage
route_delay["Delay_Frequency"] = (
    route_delay["Delay_Frequency"] * 100
)


# Sort routes from highest to lowest delay frequency
route_delay = route_delay.sort_values(
    "Delay_Frequency",
    ascending=False
)


print("\nRoute Delay Frequency:")
display(route_delay.head(20))
```

```python
# Keep only routes with at least 10 shipments
reliable_route_delay = route_delay[
    route_delay["Shipment_Volume"] >= 10
].copy()

# Sort by highest delay frequency
reliable_route_delay = reliable_route_delay.sort_values(
    "Delay_Frequency",
    ascending=False
)

print("Most Delay-Prone Reliable Routes:")
display(reliable_route_delay.head(10))
```

```python
# Combine route performance and delay information
final_route_analysis = reliable_routes.join(
    route_delay[
        ["Delayed_Shipments", "Delay_Frequency"]
    ]
)

# Sort routes by average lead time
final_route_analysis = final_route_analysis.sort_values(
    "Average_Lead_Time"
)

print("Final Route Efficiency Analysis:")
display(final_route_analysis.head(20))
```

```python
# Save the final route efficiency analysis
final_route_analysis.to_csv(
    r"D:\Nassu Project\Nassau-Candy-Shipping-Analysis\data\processed\route_efficiency_analysis.csv",
    index=True
)

print("Route efficiency analysis saved successfully.")
```
